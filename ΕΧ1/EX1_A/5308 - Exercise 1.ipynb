```python
import random
#5308 Αριστείδης Νικολακόπουλος
def sample(k,f):
    i = 0
    fix = 0
    reservoir = []
    random.seed(a = None)
    for line in open(f):
        if fix == 0:
            fix = 1
        else:
            if i <= k:
                reservoir.append(line)
            else:
                j = random.randint(0,i)
                if j<=k:
                    reservoir[j-1] = line

        i+=1

    return reservoir
```

```python
r = sample(10,"input-A.csv")

with open("sample.csv", "w") as f:
  f.write(",Player,Start Year,End Year,Position,Height,Weight,Birthday,Colleges,College Link\n")
  for line in r :
    f.write(line)
    print(line)
```

* Ο αλγόριθμος ξεκινάει με μια κενή λίστα, όπου τοποθετεί Κ αντικείμενα, και έναν μετρητή

* Kάθε n-οστο αντικείμενο , με n>Κ , επιλέγεται με πιθανότητα Κ/n και αντικαθιστά ένα τυχαίο αντικείμενο μέσα στην λίστα και ενημερώνεται πάντα ο μετρητής.

* Τέλος επιστρέφεται η λίστα με τα αντικείμενα
